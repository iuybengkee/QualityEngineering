```
from io import BytesIO
import requests
import json
import pandas as pd
from pyspark.sql.functions import lit
from pyspark.sql.types import StringType
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import datetime
from pyspark.sql.types import TimestampType, BinaryType
from pyspark.sql import functions as F
import re
```

```
class TestAutomation:
    def __init__(self, project_name, tenant_id, client_id, client_secret, tenant_name, sharepoint_site_url, sharepoint_relative_url, sql_database_url, sql_sttm_table, sql_username, sql_password, sql_template_table, sql_log_table, dim_fact_table_names, dim_fact_schema, test_case_id, storage_account_name, remarks, local_var):
        self.project_name = project_name
        self.tenant_id = tenant_id
        self.client_id = client_id
        self.client_secret = client_secret
        self.tenant_name = tenant_name
        self.sharepoint_site_url = sharepoint_site_url
        self.sharepoint_relative_url = sharepoint_relative_url
        self.sql_database_url = sql_database_url
        self.sql_sttm_table = sql_sttm_table
        self.sql_username = sql_username
        self.sql_password = sql_password
        self.sql_template_table = sql_template_table
        self.sql_log_table = sql_log_table
        self.transformed_sttm_df = None
        self.combined_df = None
        self.all_dim_fact_table_names = None
        self.dim_fact_table_names = dim_fact_table_names
        self.dim_fact_schema = dim_fact_schema
        self.test_case_id = test_case_id
        self.storage_account_name = storage_account_name
        self.remarks = remarks
        self.local_var = local_var

    def find_table_names(self, sql_query):
        """
        Find table names mentioned within double quotes in the given SQL query.

        Args:
           sql_query (str): The SQL query string to search for table names.

        Returns:
            list: A list containing the table names found within double quotes.
        """

        # Use regex to find all words within double quotes
        table_names = re.findall(r'\[([^[\]]+)\]', sql_query)
        return table_names


    def generate_dynamic_transformation_sql(self, df, SilverGoldName):

        """
        Generate dynamic transformation SQL based on the DataFrame and target table name.

        Args:
            df (DataFrame): Input DataFrame containing transformation logic in STTM.
            SilverGoldName (str): The target table name for the transformation.

        Returns:
            str: The dynamically generated SQL statement for the transformation.
        """
    
        silver_main = df.select("SourceTableName").where(F.col("SourceSQLJoin").isNull()).distinct().collect()[0][0]
        target_table_name = SilverGoldName
        table_columns = []
        transformation_clauses = []

        silver_main_path_location = df.filter((col("SourceTableName") == silver_main) & (col("BronzeLoadType").isNotNull())).select("BronzePathLocation").distinct().first()[0]

        # Loop through each row in the DataFrame to build column definitions and transformation clauses
        for row in df.collect():
            if row['SourceColumnName'] is None:
                continue
            elif row['SourceSQLJoin'] is None and row['SourceTransformLogic'] is None:
                table_columns.append(f"{row['SourceTableName']}.{row['SourceColumnName']} AS {row['ColumnName']}")
            elif row['SourceSQLJoin'] is None and row['SourceTransformLogic'] is not None:
                table_columns.append(f"{row['SourceTransformLogic']} AS {row['ColumnName']}")
            elif row['SourceSQLJoin'] is not None:
                table_columns.append(f"{row['SourceTransformLogic']} AS {row['ColumnName']}")
                transformation_clauses.append(f"{row['SourceSQLJoin']}")

        # Join the column definitions with commas to create the SELECT statement
        columns_str = ',\n'.join(table_columns)
        
        # Remove duplicates from the transformation clauses list
        transformation_clauses = list(set(transformation_clauses))
        # Join the transformation clauses to create the additional SQL transformations
        transformations_str = '\n'.join(transformation_clauses)


        # Use .format() to replace the placeholders in the SQL statement and create the final SQL query
        sql_statement = """SELECT
{}
FROM delta.`abfss://silver-raw@{}.dfs.core.windows.net/{}/{}` AS {}
{}""".format(columns_str, self.storage_account_name, silver_main_path_location, silver_main, silver_main, transformations_str)

        # Find all words within double quotes mentioned in the SQL statement.
        # These words within double quotes represent table names in the SQL statement.
        # Then replace each table name with the full path of that table in the data lake. 
        table_names = self.find_table_names(sql_statement)

        for table_name in table_names:
            path_location = df.filter(col("SourceTableName") == table_name).select("BronzePathLocation").distinct().first()[0]
            table_name_without_quotes = table_name.replace('[', '')
            table_name_without_quotes = table_name.replace(']', '')
            sql_statement = re.sub(rf'"{table_name_without_quotes}"', f"delta.`abfss://silver-raw@{self.storage_account_name}.dfs.core.windows.net/{path_location}/{table_name_without_quotes}`", sql_statement)

        return sql_statement


    def build_properties(self, sql_username, sql_password):

        
        """
        Build and return the properties dictionary required for connecting to a SQL Server database

        Args:
            sql_username (str): The username for the SQL Server connection.
            sql_password (str): The password for the SQL Server connection.

        Returns:
            dict: A dictionary containing the necessary properties for the SQL Server connection.
        """

        properties = {
          "user": sql_username,
          "password": sql_password,
          "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
          }

        return properties

    def write_to_sql_db(self, df, sql_database_url, sql_log_table, sql_username, sql_password, writeMode):
        
        """
        Write the log results to a specified SQL database table.

        Args:
            log_df (DataFrame): The DataFrame containing the data to be written to the SQL database.
            sql_database_url (str): The URL or connection string for the SQL database.
            sql_log_table (str): The name of the target table in the SQL database.
            sql_username (str): The username for the SQL database connection.
            sql_password (str): The password for the SQL database connection.
            writeMode (str): The mode for writing the data to the SQL database (e.g., "overwrite", "append", "ignore", etc.).

        Returns:
            None
        """
        df.write \
            .format("jdbc") \
            .option("url", sql_database_url) \
            .option("dbtable", sql_log_table) \
            .option("user", sql_username) \
            .option("password", sql_password) \
            .mode(writeMode) \
            .save()        

    def log_test_case_result(self, table_name, test_case_id, table_type, expected_value, actual_value, test_result, test_case_script, test_case_name, add_remarks=None):

        """
        Log the result of a test case to the SQL database.

        Args:
            table_name (str): The name of the table being tested.
            test_case_id (int): The unique identifier of the test case.
            table_type (str): The type of the table (e.g., "fact" or "dimension").
            test_result_count (int): The count of test results.
            test_result (str): The outcome of the test case (e.g., "passed" or "failed").

        Returns:
            None
        """

        current_time = datetime.datetime.now()

        if add_remarks:
            remarksEdited = self.remarks + " - " + str(add_remarks)
        else:
            remarksEdited = self.remarks
            
        # Prepare the log data for the test case.
        log_data = [(current_time, test_case_id, self.project_name, table_type, "EDW", table_name, test_case_name, test_case_script, expected_value, actual_value, test_result, remarksEdited)]

        # Create a DataFrame for the log data.
        log_df = spark.createDataFrame(log_data, ["DateTested", "TestCaseId", "ProjectName", "TableType", "TableSchema", "TableName", "TestCaseName", "TestCaseScript", "ExpectedValue", "ActualValue", "TestResult", "Remarks"])

        # Write the log DataFrame to the SQL database.
        self.write_to_sql_db(log_df, self.sql_database_url, self.sql_log_table, self.sql_username, self.sql_password, "append")


    def get_token(self):
        
        """
        Obtain an authentication token for SharePoint using the provided credentials.

        Returns:
            dict: A dictionary containing the authentication token and related information.
        """

        # Prepare the data with necessary credentials to request the authentication token.
        data = {
            'grant_type': 'client_credentials',
            'resource': "00000003-0000-0ff1-ce00-000000000000/" + self.tenant_name + ".sharepoint.com@" + self.tenant_id,
            'client_id': self.client_id + '@' + self.tenant_id,
            'client_secret': self.client_secret,
        }
        headers = {'Content-Type': 'application/x-www-form-urlencoded'}
        
        # Prepare the URL for the authentication request.
        authUrl = f"https://accounts.accesscontrol.windows.net/{self.tenant_id}/tokens/OAuth/2"
        
        # Send a POST request to obtain the authentication token.
        response = requests.post(authUrl, data=data, headers=headers)

        # Parse the response text containing the token information into a dictionary.
        return json.loads(response.text)

    def get_files(self):

        """
        Get a list of files from the SharePoint site.

        Returns:
            list: A list of dictionaries, each containing file metadata like 'Name' and 'ServerRelativeUrl'.
        """

        # Initialize an empty list to store file metadata.
        files = []

        # Obtain the access token using the provided credentials.
        token_val = self.get_token()['access_token']
        
        # Prepare the headers for SharePoint API request.
        headers = {
            'Authorization': "Bearer " + token_val,
            'Accept': 'application/json;odata=verbose',
            'Content-Type': 'application/json;odata=verbose'
        }
        
        # Prepare the URL for the SharePoint API request to get files.
        responseUrl = f"{self.sharepoint_site_url}/_api/web/GetFolderByServerRelativeUrl('{self.sharepoint_relative_url}')/Files"
        
        # Send a GET request to retrieve file information from SharePoint.
        response = requests.get(responseUrl, headers=headers)
        
        # Parse the JSON response from SharePoint into a dictionary.
        json_output = json.loads(response.text)

        # Extract the file metadata from the JSON response and add them to the 'files' list.
        for fl in json_output.get('d').get('results'):
            files.append({'Name': fl.get('Name'), 'ServerRelativeUrl': fl.get('ServerRelativeUrl')})

        # Return the list of files with their metadata.
        return files

    def download_files(self, file_path):
        
        """
        Download a file from the SharePoint site.

        Args:
            file_path (str): The server-relative path of the file to be downloaded.

        Returns:
            bytes: The content of the downloaded file as bytes.
        """
        
        # Prepare the URL for the SharePoint API request to download the file.
        FileUrl = f"{self.sharepoint_site_url}/_api/web/getFileByServerRelativePath(DecodedUrl='{file_path}')/\$value"
        
        # Obtain the access token using the provided credentials.
        token_val = self.get_token()['access_token']
        
        # Prepare the headers for the SharePoint API request.
        headers = {
            'Authorization': "Bearer " + token_val,
            'Accept': 'application/json;odata=verbose',
            'Content-Type': 'application/json;odata=verbose'
        }
        
        # Send a GET request to download the file from SharePoint.
        response = requests.get(FileUrl, headers=headers)

        # Return the content of the downloaded file as bytes.
        return response.content

    def num_duplicate_keys(self, df, column_name):
        total_count = df.count()
        distinct_count = df.select(*column_name).distinct().count()
        return total_count - distinct_count

    def check_columns_match(self, df1, df2):
        """
        Check if the column names and data types between two DataFrames match.
        Args:
            df1 (DataFrame): First DataFrame.
            df2 (DataFrame): Second DataFrame.

        Returns:
            bool: True if the column names and data types match, False otherwise.
        """
        return df1.schema == df2.schema

    def column_name_list_sort(self, column_name_list):
        """
        Retruns a sorted list of column names based on column names in lower case
        """
        return sorted(column_name_list, key = lambda element : element.lower())


    def read_sttm_from_sharepoint(self):

        """
        Process the files obtained from SharePoint and store the data in the SQL database.

        Returns:
            None
        """

        # Get the list of files from the SharePoint site.
        file_list = self.get_files()

        # Initialize a DataFrame to combine the data from all files.
        combined_df = None

        # Loop through each file and process its content.
        for file in file_list:

            # Download the file content from SharePoint and create a DataFrame.
            response_content = self.download_files(file.get('ServerRelativeUrl'))

            # Convert the downloaded file content into a bytes-like object.
            bytes_file_obj = BytesIO(response_content)

            # Create a pandas ExcelFile object using the bytes-like object to read the Excel file content.
            xls = pd.ExcelFile(bytes_file_obj)
            
            if 'all' in [table.lower().strip() for table in self.dim_fact_table_names]:
                self.all_dim_fact_table_names = [sheet for sheet in xls.sheet_names if sheet.lower().startswith("dim") or sheet.lower().startswith("fact")]
            else:
                self.all_dim_fact_table_names = self.dim_fact_table_names

            # Read each sheet in the ExcelFile object `xls` into a list of pandas DataFrames
            df_list = [pd.read_excel(xls, sheet_name=sheet, dtype=str) for sheet in self.all_dim_fact_table_names if sheet in xls.sheet_names]

            # Concatenate all the DataFrames in the `df_list` into a single combined pandas DataFrame
            combined_pd_df = pd.concat(df_list, ignore_index=True)

            # Create a Spark DataFrame `spark_df` from the combined pandas DataFrame `combined_pd_df`.
            spark_df = spark.createDataFrame(combined_pd_df)

            # Merge the current DataFrame with the previously processed files.
            if combined_df is None:
                combined_df = spark_df
            else:
                combined_df = combined_df.union(spark_df)

        # Perform necessary transformations and renaming on the combined DataFrame.
        self.transformed_sttm_df = combined_df.withColumn('ProjectName', lit(self.project_name).cast(StringType())) \
            .withColumnRenamed('TableSchemaName', 'TableSchema') \
            .withColumnRenamed('Parent.Mapped to', 'SourceTableName') \
            .withColumnRenamed('Display Name', 'SourceColumnName') \
            .withColumnRenamed('Mapped to column', 'SourceTransformLogic') \
            .withColumnRenamed('Table', 'TableName') \
            .withColumnRenamed('Column.Name', 'ColumnName') \
            .withColumnRenamed('Target Data Type', 'ColumnDataType') \
            .withColumnRenamed('Source Precision', 'Precision').withColumn('Precision', col('Precision').cast(IntegerType())) \
            .withColumnRenamed('Source Length', 'Length').withColumn('Length', col('Length').cast(IntegerType())) \
            .withColumnRenamed('Mapped to join', 'SourceSQLJoin') \
            .withColumnRenamed('Creation Date', 'CreationDateTime').withColumn('CreationDateTime', col('CreationDateTime').cast(TimestampType())) \
            .withColumnRenamed('Modification Date', 'ModificationDateTime').withColumn('ModificationDateTime', col('ModificationDateTime').cast(TimestampType()))

        if "SCDType" in combined_df.columns:
            # If "SCDType" column exists, rename it to 'SCD_type'
            self.transformed_sttm_df = self.transformed_sttm_df.withColumnRenamed('SCDType', 'SCD_type')
        else:
            # If "SCDType" column does not exist, add a new column 'SCD_type' with a null value
            self.transformed_sttm_df = self.transformed_sttm_df.withColumn('SCD_type', expr('CAST(NULL AS STRING)'))

        # Define the list of columns to check for existence and add if missing
        columns_to_check = ["IS_NULLABLE", "MandatoryTechnicalFIeld", "DistributionBy", "PartitionBy", "ColumnDefaultValue", "BronzeLoadType"]

        # Add missing columns with null values
        for column in columns_to_check:
            if column not in combined_df.columns:
                self.transformed_sttm_df = self.transformed_sttm_df.withColumn(column, expr('CAST(NULL AS STRING)'))

        # Select only the required columns from the DataFrame.
        self.transformed_sttm_df = self.transformed_sttm_df.select(
            'ProjectName',
            'TableType',
            'TableSchema',
            'TableName',
            'ColumnName',
            'ColumnDataType',
            'Precision',
            'Length',
            'MandatoryTechnicalFIeld',
            'ColumnType',
            'IS_NULLABLE',
            'SCD_type',
            'DistributionBy',
            'PartitionBy',
            'ParentTableName',
            'ParentColumnName',
            'ColumnDefaultValue',
            'BronzeLoadType',
            'BronzeFileFormat',
            'BronzePathLocation',
            'BronzeTableName',
            'SourceLocation',
            'SourceTableName',
            'SourceColumnName',
            'SourceTransformLogic',
            'SourceSQLJoin',  
            'Comment',
            'CreationDateTime',
            'Creator',
            'ModificationDateTime',
            'Modifier'
            )

    def write_sttm_into_sql_db(self):
        
        """
        Write the processed DataFrame into the SQL database.

        This function first truncates the STTM table by writing an empty DataFrame with the same schema as the SQL table.
        Truncating helps prevent unintended changes to the table structure during an overwrite operation.

        The processed DataFrame is then appended to the STTM table in the SQL database.

        Returns:
            None
        """

         # Write an empty DataFrame with the same schema as the SQL table to truncate the table.
        #properties = self.build_properties(self.sql_username, self.sql_password)
        schema = spark.read.jdbc(url=sql_database_url, table=self.sql_sttm_table, properties=properties).schema
        empty_df = spark.createDataFrame([], schema)
        self.write_to_sql_db(empty_df, self.sql_database_url, self.sql_sttm_table, self.sql_username, self.sql_password, "overwrite")

        int_columns = ["Precision", "Length"]
        timestamp_columns = ["CreationDateTime", "ModificationDateTime"]

        # Handle null columns to prevent failure
        self.transformed_sttm_df = self.transformed_sttm_df.select([when(col(c).isNull() & (col(c).isin(int_columns)), col(c).cast(IntegerType())) \
                                                    .when(col(c).isNull() & (col(c).isin(timestamp_columns)), col(c).cast(TimestampType())) \
                                                    .otherwise(col(c).cast(StringType())) \
                                                    .alias(c) \
                                                    for c in self.transformed_sttm_df.columns])

        self.write_to_sql_db(self.transformed_sttm_df, self.sql_database_url, self.sql_sttm_table, self.sql_username, self.sql_password, "append")

    def BronzeSilverTest(self):

        for gold_table in self.all_dim_fact_table_names:
            #display(self.transformed_sttm_df)
            filtered_sttm_df = self.transformed_sttm_df.where(col("TableName") == gold_table)
            distinct_tables = filtered_sttm_df.select("SourceTableName").distinct().collect()

            if distinct_tables[0]['SourceTableName']:
                for item in distinct_tables:

                    filtered_sttm_BronzePathLocation_df = filtered_sttm_df.filter((col("SourceTableName") == item['SourceTableName']) & (col("BronzePathLocation").isNotNull()))
                    if filtered_sttm_BronzePathLocation_df.isEmpty():
                        self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"BronzePathLocation not defined in STTM for {gold_table}!")
                        raise ValueError(f"BronzePathLocation not defined in STTM for {gold_table}!")
                    else:
                        path_location = filtered_sttm_BronzePathLocation_df.select("BronzePathLocation").distinct().first()[0]


                    filtered_sttm_BronzeLoadType_df = filtered_sttm_df.filter((col("SourceTableName") == item['SourceTableName']) & (col("BronzeLoadType").isNotNull()))
                    if filtered_sttm_BronzeLoadType_df.isEmpty():
                        self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"BronzeLoadType not defined in STTM for {gold_table}!")
                        raise ValueError(f"BronzeLoadType not defined in STTM for {gold_table}!")
                    else:
                        bronze_load_type = filtered_sttm_BronzeLoadType_df.select("BronzeLoadType").distinct().first()[0]


                    filtered_sttm_PrimaryKeyColumn_df = filtered_sttm_df.filter((col("SourceTableName") == item['SourceTableName']) & (col("ColumnType") == "BK"))
                    if filtered_sttm_PrimaryKeyColumn_df.isEmpty():
                        self.log_test_case_result(gold_table, 0, "Silver", 0, 1, "FAIL", "NA", "NA", f"PrimaryKeyColumn not defined in STTM for {gold_table}!")
                        raise ValueError(f"PrimaryKeyColumn not defined in STTM for {gold_table}!")
                    else:
                        primary_key_column = filtered_sttm_PrimaryKeyColumn_df.select("SourceColumnName").collect()
                        primary_key_column = [row[0] for row in primary_key_column]
                        primary_key_column_str = ", ".join([f"'{col}'" for col in primary_key_column])
                   

                    filtered_sttm_BronzeFileFormat_df = filtered_sttm_df.filter((col("SourceTableName") == item['SourceTableName']) & (col("BronzeFileFormat").isNotNull()))
                    if filtered_sttm_BronzeFileFormat_df.isEmpty():
                        self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"BronzeFileFormat not defined in STTM for {gold_table}!")
                        raise ValueError(f"BronzeFileFormat not defined in STTM for {gold_table}!")
                    else:
                        bronze_file_format = filtered_sttm_BronzeFileFormat_df.select("BronzeFileFormat").distinct().first()[0]


                    # Base path containing the nested directories
                    bronze_base_path = f"abfss://bronze@{self.storage_account_name}.dfs.core.windows.net/{path_location}/{item['SourceTableName']}"

                    if bronze_load_type.lower() == "full":    
                        # Get all directories in the base path
                        bronze_directories = [file_info.path for file_info in dbutils.fs.ls(bronze_base_path) if file_info.isDir()]

                        if bronze_directories:
                            # Select the last date directory (latest timestamp)
                            bronze_last_date_dir = bronze_directories[-1]

                            # Get all directories within the last date directory
                            bronze_subdirectories = [file_info.path for file_info in dbutils.fs.ls(bronze_last_date_dir) if file_info.isDir()]

                            if bronze_subdirectories:
                                # Select the last time directory inside the last date directory (latest timestamp)
                                bronze_last_time_dir_inside_last_date = bronze_subdirectories[-1]

                                # Read the last CSV/parquet file in the last time directory
                                if bronze_file_format.lower() == 'csv':
                                    Bronze_df = spark.read.option('inferSchema', 'true').option('header', 'true').csv(bronze_last_time_dir_inside_last_date)  
                                elif bronze_file_format.lower() == 'parquet':
                                    Bronze_df = spark.read.parquet(bronze_last_time_dir_inside_last_date)
                                else:
                                    self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"{bronze_file_format} not valid for {item['SourceTableName']}.")
                                    raise ValueError(f"{bronze_file_format} is not a valid FileFormat in STTM for {item['SourceTableName']}!")

                            else:
                                self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"No time directory found for {item['SourceTableName']}.")
                                raise FileNotFoundError(f"No time directory found for {item['SourceTableName']}.")
                        else:
                            self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"No directories with date and time layers found for {item['SourceTableName']}.")
                            raise FileNotFoundError(f"No directories with date and time layers found for {item['SourceTableName']}.")

                    elif bronze_load_type.lower() == "incremental":
                        bronze_base_path_incremental = f"abfss://bronze@{self.storage_account_name}.dfs.core.windows.net/{path_location}/{item['SourceTableName']}" + "/*/*/"
                        #Bronze_df = spark.read.schema(getSchema).option("recursive", "true").option('header', 'true').csv(bronze_base_path)
                        if bronze_file_format.lower() == 'csv':
                            Bronze_df = spark.read.option('inferSchema', 'true').option("recursive", "true").option('header', 'true').csv(bronze_base_path_incremental)
                        elif bronze_file_format.lower() == 'parquet':
                            Bronze_df = spark.read.option("recursive", "true").parquet(bronze_base_path_incremental)
                        else:
                            self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"{bronze_file_format} not valid for {item['SourceTableName']}.")
                            raise ValueError(f"{bronze_file_format} is not a valid FileFormat in STTM for {item['SourceTableName']}!")

                    else:
                        self.log_test_case_result(gold_table, 0, "Bronze", 0, 1, "FAIL", "NA", "NA", f"{bronze_load_type} loadType is not valid!")
                        raise ValueError(f"{bronze_load_type} loadType is not valid!")

                    # Base path containing the nested directories
                    silver_base_path = f"abfss://silver-raw@{self.storage_account_name}.dfs.core.windows.net/{path_location}/{item['SourceTableName']}"

                    if not dbutils.fs.ls(silver_base_path):
                        self.log_test_case_result(gold_table, 0, "Silver", 0, 1, "FAIL", "NA", "NA", f"Delta Lake files not found at the specified path: {silver_base_path}")
                        raise FileNotFoundError(f"Delta Lake files not found at the specified path: {silver_base_path}")


                    Silver_df = spark.read.format("delta").load(silver_base_path).where("`__current` = true and `__deleted` = false")


                    # Test -1
                    if '-1' in self.test_case_id or 'all' in self.test_case_id:

                        
                        test1_script = f"SELECT CASE WHEN COUNT(*) > 0 THEN 1 ELSE 0 END AS silver_count FROM delta.`{silver_base_path}`"

                        silver_count = 1 if Silver_df.count() > 0 else 0

                        if silver_count == 1:                
                            self.log_test_case_result(gold_table, -1, "Silver", 1, 1, "PASS",test1_script,"Validate Table Row Count Silver", ("SilverCount: "+str(silver_count)))
                        else:
                            self.log_test_case_result(gold_table, -1, "Silver", 1, 0, "FAIL",test1_script, "Validate Table Row Count Silver", ("SilverCount: "+str(silver_count)))
                        print("Test -1 completed!")

                    # Test -2
                    if '-2' in self.test_case_id or 'all' in self.test_case_id:
                        
                        test2_script = f"""SELECT (SELECT COUNT(DISTINCT concat({primary_key_column_str})) FROM {item['SourceTableName']}) AS bronze_count - (SELECT COUNT(DISTINCT concat({primary_key_column_str})) FROM delta.`{silver_base_path}`) AS silver_count"""

                        # Query to get the count of distinct primary keys
                        bronze_count = Bronze_df.select(primary_key_column).distinct().count()
                        silver_count = Silver_df.select(primary_key_column).distinct().count()

                        if bronze_count == silver_count:
                            self.log_test_case_result(gold_table, -2, "Bronze/Silver", 0, 0, "PASS", test2_script, "Validate Row Count against Bronze", None)
                        else:
                            self.log_test_case_result(gold_table, -2, "Bronze/Silver", 0, (bronze_count-silver_count), "FAIL", test2_script, "Validate Row Count against Bronze", None)
                        print("Test -2 completed!")

                    # Test -3
                    if '-3' in self.test_case_id or 'all' in self.test_case_id:

                        test3_script = f"""SELECT COUNT(1) FROM ( SELECT COUNT(1) AS cnt, {primary_key_column_str} FROM delta.`{silver_base_path}` where 
                        --current = 1 and __deleted = 0 GROUP BY {primary_key_column_str} HAVING COUNT(1) > 1 )"""

                        # Check for duplicate keys in the Silver DataFrame
                        silver_duplicates = self.num_duplicate_keys(Silver_df, primary_key_column)
                        
                        if silver_duplicates == 0:
                            self.log_test_case_result(gold_table, -3, "Silver", 0, 0, "PASS", test3_script, "Validate Deduplicate Row", None)
                        else:
                            self.log_test_case_result(gold_table, -3, "Silver", 0, silver_duplicates, "FAIL", test3_script, "Validate Deduplicate Row",None)
                        print("Test -3 completed!")
            else:
                self.log_test_case_result(gold_table, -1, "Bronze/Silver", 0, 1, "FAIL", "NA", f"No source tables found in STTM for {gold_table}!")
                raise ValueError(f"No source tables found in STTM for {gold_table}!")

    def SilverGoldTest(self):
        for gold_table in self.all_dim_fact_table_names:
            # Filter the transformed_sttm_df to get the specific table data
            filtered_sttm_gold_df = self.transformed_sttm_df.where(col("TableName") == gold_table)
            

            if f"{gold_table}_ST" in self.local_var:
                sql_transformation = self.local_var[f"{gold_table}_ST"]
            else:
                # Generate the dynamic SQL transformation
                sql_transformation = self.generate_dynamic_transformation_sql(filtered_sttm_gold_df, gold_table)
                
            # Execute the SQL query and get the expected_df
            expected_df = spark.sql(sql_transformation)
            expected_df = expected_df.select(*[col(c) for c in expected_df.columns if not c.startswith("__") and not c.endswith("SID")])

            # Read the target DataFrame directly from the database and select relevant columns
            properties = self.build_properties(self.sql_username, self.sql_password)
            target_df = spark.read.jdbc(url=self.sql_database_url, table= f'{self.dim_fact_schema}.{gold_table}', properties=properties)
            target_without_audit_df = target_df.select(*[col(c) for c in target_df.columns if not c.startswith("__") and not c.endswith("SID")])


            filtered_fact_dim_table_type_df = filtered_sttm_gold_df.filter(col("TableType").isNotNull())
            if filtered_fact_dim_table_type_df.isEmpty():
                self.log_test_case_result(gold_table, 0, "Gold", 0, 1, "FAIL", "NA", "NA",f"TableType not defined in STTM for {gold_table}!")
                raise ValueError(f"TableType not defined in STTM for {gold_table}!")
            else:
                fact_dim_table_type = filtered_fact_dim_table_type_df.select("TableType").distinct().first()[0]
            

            if '13' in self.test_case_id or 'all' in self.test_case_id:
                

                expected_columns = sorted(expected_df.columns)
                target_columns = sorted(target_without_audit_df.columns)

                expected_df=expected_df.selectExpr(*expected_columns)
                target_without_audit_df=target_without_audit_df.selectExpr(*target_columns)

                expected_columns_str = ", ".join([f"'{col}'" for col in expected_columns])
                target_columns_str = ", ".join([f"'{col}'" for col in target_columns])
                test13_script = f"""SELECT COUNT(1) FROM (SELECT {expected_columns_str} FROM ({sql_transformation}) AS SOURCE EXCEPT SELECT {target_columns_str} FROM {self.dim_fact_schema}.{gold_table} ) a"""
                
                
                if expected_columns == target_columns:
                    # Compare the two DataFrames and get the count of rows with differences
                    row_differences = expected_df.exceptAll(target_without_audit_df).count()

                    if row_differences == 0:
                        self.log_test_case_result(gold_table, 13, fact_dim_table_type.capitalize(), 0, 0, "PASS", test13_script, "Validate Minus Query Silver-RAW to Gold",None)
                    else:
                        self.log_test_case_result(gold_table, 13, fact_dim_table_type.capitalize(), 0, row_differences, "FAIL", test13_script,"Validate Minus Query Silver-RAW to Gold", None)                    
                        
                else:
                    self.log_test_case_result(gold_table, 13, fact_dim_table_type.capitalize(), 0,1, "FAIL", test13_script, "Validate Minus Query Silver-RAW to Gold","The expected columns do not align with the columns in target table!")
                
                print("Test 13 completed!")

            if fact_dim_table_type.lower() == 'dim':

                if '16' in self.test_case_id or 'all' in self.test_case_id:


                    # Filter the DataFrame to include only rows where SCDType is "Type1"
                    type1_filtered_df = filtered_sttm_gold_df.filter(col('SCD_type') == 'Type 1')

                    # Collect the column names from the filtered DataFrame
                    type1_column_names_list = type1_filtered_df.select('ColumnName').rdd.flatMap(lambda x: x).collect()
                    
                    if type1_column_names_list:
                        if set(type1_column_names_list).issubset(target_df.columns):
                            
                            type1_column_names_list_sorted = self.column_name_list_sort(type1_column_names_list)
                            df_with_hash = target_df.withColumn('CalculatedType1Hash', sha2(concat_ws('|', *[col(column).cast(StringType()) for column in  type1_column_names_list_sorted]), 256).cast(BinaryType()))
                            type1_count = df_with_hash.select('CalculatedType1Hash').subtract(target_df.select('__Type1Hash')).count()

                            type1_column_names_list_sorted_concat = ', '.join([f"{col}::STRING" for col in type1_column_names_list_sorted])
                            test16_script = f"""SELECT count(1) FROM (SELECT SHA2(CONCAT_WS('|', {type1_column_names_list_sorted_concat}), 256) AS CalculatedType1Hash from {self.dim_fact_schema}.{gold_table}) SubQuery WHERE CalculatedType1Hash NOT IN (SELECT __Type1Hash from {self.dim_fact_schema}.{gold_table})"""

                            if type1_count == 0:
                                self.log_test_case_result(gold_table, 16, "Dim", 0,0, "PASS", test16_script, "Validate Type1Hash",None)
                            else:
                                self.log_test_case_result(gold_table, 16, "Dim", 0,type1_count, "FAIL", test16_script, "Validate Type1Hash", "Type1Hash columns do not match!")
                        else:
                            self.log_test_case_result(gold_table, 16, "Dim", 0,1, "FAIL", "NA", "Validate Type1Hash","Some Type 1 columns in STTM do not exist in the target table!")
                    else:
                        self.log_test_case_result(gold_table, 16, "Gold", 0, 1, "FAIL", "NA", "NA", f"No Type 1 column found in STTM for {gold_table}!")
                    
                    print("Test 16 completed!")

                if '17' in self.test_case_id or 'all' in self.test_case_id:

                    # Filter the DataFrame to include only rows where SCDType is "Type2"
                    type2_filtered_df = filtered_sttm_gold_df.filter(col('SCD_type') == 'Type 2')

                    # Collect the column names from the filtered DataFrame
                    type2_column_names_list = type2_filtered_df.select('ColumnName').rdd.flatMap(lambda x: x).collect()


                    if type2_column_names_list:
                        if set(type2_column_names_list).issubset(target_df.columns):
                            
                            type2_column_names_list_sorted = self.column_name_list_sort(type2_column_names_list)
                            df_with_hash = target_df.withColumn('CalculatedType2Hash', sha2(concat_ws('|', *[col(column).cast(StringType()) for column in  type2_column_names_list_sorted]), 256).cast(BinaryType()))
                            type2_count = df_with_hash.select('CalculatedType2Hash').subtract(target_df.select('__Type2Hash')).count()

                            type2_column_names_list_sorted_concat = ', '.join([f"{col}::STRING" for col in type2_column_names_list_sorted])
                            test17_script = f"""SELECT count(1) FROM (SELECT SHA2(CONCAT_WS('|', {type2_column_names_list_sorted_concat}), 256) AS CalculatedType2Hash from {self.dim_fact_schema}.{gold_table}) SubQuery WHERE CalculatedType1Hash NOT IN (SELECT __Type2Hash from {self.dim_fact_schema}.{gold_table})"""

                            if type2_count == 0:
                                self.log_test_case_result(gold_table, 17, "Dim", 0,0, "PASS", test17_script, "Validate Type2Hash", None)
                            else:
                                self.log_test_case_result(gold_table, 17, "Dim", 0,type2_count, "FAIL", test17_script, "Validate Type2Hash","Type2Hash columns do not match!")
                        else:
                            self.log_test_case_result(gold_table, 17, "Dim", 0,1, "FAIL", "NA", "Validate Type2Hash","Some Type 2 columns in STTM do not exist in the target table!")
                    else:
                        self.log_test_case_result(gold_table, 17, "Gold", 0, 1, "FAIL", "NA","NA", f"No Type 2 column found in STTM for {gold_table}!")

                    print("Test 17 completed!")
```
